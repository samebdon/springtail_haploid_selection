```python
#folder = './all/'
folder = './sco/'
```

```python
obs_A_alpha = folder+'obs_A_alpha.txt'
obs_X_alpha = folder+'obs_X_alpha.txt'
obs_A_dfe = folder+'obs_A_dfe.txt'
obs_X_dfe = folder+'obs_X_dfe.txt'

bootstrap_A_alpha = folder+'bootstrap_A_alpha.txt'
bootstrap_X_alpha = folder+'bootstrap_X_alpha.txt'
bootstrap_A_dfe = folder+'bootstrap_A_dfe.txt'
bootstrap_X_dfe = folder+'bootstrap_X_dfe.txt'
```

```python
obs_A_alpha_df = pd.read_csv(obs_A_alpha, sep=' ', header=None)
obs_X_alpha_df = pd.read_csv(obs_X_alpha, sep=' ', header=None)

obs_A_dfe_df = pd.read_csv(obs_A_dfe, sep=' ', header=None, usecols=[0,5,9,13])
obs_X_dfe_df = pd.read_csv(obs_X_dfe, sep=' ', header=None, usecols=[0,5,9,13])

bootstrap_A_alpha_df = pd.read_csv(bootstrap_A_alpha, sep=' ', header=None)
bootstrap_A_alpha_df[1] = pd.to_numeric(bootstrap_A_alpha_df[1])
bootstrap_A_alpha_df[3] = pd.to_numeric(bootstrap_A_alpha_df[3])
bootstrap_X_alpha_df = pd.read_csv(bootstrap_X_alpha, sep=' ', header=None)
bootstrap_X_alpha_df[1] = pd.to_numeric(bootstrap_X_alpha_df[1])
bootstrap_X_alpha_df[3] = pd.to_numeric(bootstrap_X_alpha_df[3])

bootstrap_A_dfe_df = pd.read_csv(bootstrap_A_dfe, sep=' ', header=None, usecols=[0,5,9,13])
bootstrap_X_dfe_df = pd.read_csv(bootstrap_X_dfe, sep=' ', header=None, usecols=[0,5,9,13])
```

```python
def get_alpha_stats(obs, bootstrap):
    obs_alpha = obs[1][0]
    obs_omega_alpha = obs[3][0]
    
    df = bootstrap[1].describe()[["count", "mean", "std"]]
    alpha_upper = obs_alpha + 1.96*(df["std"]/np.sqrt(df["count"]))
    alpha_lower = obs_alpha - 1.96*(df["std"]/np.sqrt(df["count"]))
    
    df = bootstrap[3].describe()[["count", "mean", "std"]]
    omega_alpha_upper = obs_omega_alpha + 1.96*(df["std"]/np.sqrt(df["count"]))
    omega_alpha_lower = obs_omega_alpha - 1.96*(df["std"]/np.sqrt(df["count"]))
    
    print(f'alpha = {obs_alpha:.5f} [{alpha_lower:.5f}, {alpha_upper:.5f}]')
    print(f'omega alpha = {obs_omega_alpha:.5f} [{omega_alpha_lower:.5f}, {omega_alpha_upper:.5f}]')
    
def get_dfe_stats(obs, bootstrap):
    obs_one = obs[0][0]
    obs_two = obs[5][0]
    obs_three = obs[9][0]
    obs_four = obs[13][0]
    
    df = bootstrap[0].describe()[["count", "mean", "std"]]
    one_upper = obs_one + 1.96*(df["std"]/np.sqrt(df["count"]))
    one_lower = obs_one - 1.96*(df["std"]/np.sqrt(df["count"]))
    
    df = bootstrap[5].describe()[["count", "mean", "std"]]
    two_upper = obs_two + 1.96*(df["std"]/np.sqrt(df["count"]))
    two_lower = obs_two - 1.96*(df["std"]/np.sqrt(df["count"]))
    
    df = bootstrap[9].describe()[["count", "mean", "std"]]
    three_upper = obs_three + 1.96*(df["std"]/np.sqrt(df["count"]))
    three_lower = obs_three - 1.96*(df["std"]/np.sqrt(df["count"]))
    
    df = bootstrap[13].describe()[["count", "mean", "std"]]
    four_upper = obs_four + 1.96*(df["std"]/np.sqrt(df["count"]))
    four_lower = obs_four - 1.96*(df["std"]/np.sqrt(df["count"]))
    
    print(f"[0-1) = {obs_one:.5f} [{one_lower:.5f}, {one_upper:.5f}]")
    print(f"[1-10) = {obs_two:.5f} [{two_lower:.5f}, {two_upper:.5f}]")
    print(f"[10-100) = {obs_three:.5f} [{three_lower:.5f}, {three_upper:.5f}]")
    print(f">=100 = {obs_four:.5f} [{four_lower:.5f}, {four_upper:.5f}]")
```

```python
print('Autosomes')
get_alpha_stats(obs_A_alpha_df, bootstrap_A_alpha_df)
print('X')
get_alpha_stats(obs_X_alpha_df, bootstrap_X_alpha_df)
```

```python
print('Autosomes')
get_dfe_stats(obs_A_dfe_df, bootstrap_A_dfe_df)
print('X')
get_dfe_stats(obs_X_dfe_df, bootstrap_X_dfe_df)
```
